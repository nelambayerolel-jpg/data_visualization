```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.arange(1,12,2)
y = np.arange(2,13,2)
# create a line plot with a dashdot linestyle
plt.plot(x,y , linestyle='-.', marker ='*', markersize='20')
plt.show()
```

```python
fig ,ax = plt.subplots()
ax.plot(x, y , linestyle='-.', marker='*')
plt.show()
```

```python
plt.plot(x,y,linestyle=':')
plt.show()
```

```python

y = np.array([3, 8, 1, 10, 4, 200, 45, 67, 56, 98, 100])
```

```python
# create a dashed line , with diamond markers
plt.plot(y , linestyle='--', marker ='D', )

plt.show
```

```python
#Using the opposite approach , linestyle of dasdot , and square markers
fig , ax = plt.subplots()
ax.plot(y , linestyle='-.' , marker='s')
plt.show()
```

```python
# diamond marker ,red , marker color , marker widt
plt.plot(y ,  marker ='D',markerfacecolor ='red', markersize='7' , markeredgewidth=3)
plt.show()
```

```python

x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 170, 190, 160, 300, 310, 320, 330])
```

```python
# create a lineplot , custom color , dooted linestyle , dimond markers
plt.plot(x, y , linestyle =':', marker ='D', color=(0.4 ,0.3,0.8))

plt.title('Sports Watch Data')
plt.xlabel('Average Pulse')
plt.ylabel ('Calorie Burnage')
plt.show()
```

```python
# create a lineplot , custom color , dooted linestyle , dimond markers
plt.plot(x, y , linestyle =':', marker ='D', color='#FF5733')

plt.title('Sports Watch Data', color = 'blue', size ='18' , style ='italic')
plt.xlabel('Average Pulse', color ='green')
plt.ylabel ('Calorie Burnage', color ='red')
plt.show()
```

```python
years=[1970,1980,1985,1990,1995,2000,2005,2010,2015,2020,2025,2030,2035]
pops=[3.6,4.0,4.4,4.8,5.3,5.7,6.1,6.5,7.3,7.3,7.3,7.4,7.5]
```

```python
# title of 'World Population' - blue
# Labels growth by year , population in billions , green red
plt.plot(years, pops, linestyle ='--', marker ='*', markersize=10 , markerfacecolor='orange')

plt.title('World population' , color = 'blue')
plt.xlabel('growth by year' ,color = 'green')
plt.ylabel ('population in billions', color = 'red')
plt.show()
```
