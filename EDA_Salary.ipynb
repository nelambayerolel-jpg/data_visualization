Taking a closer look at education and finalweight

. goal: is toobserve the amount of the population that of observed in each education level

```python
df['education'].value_counts()
```

```python
subset = df.groupby('education')['fnlwgt'].sum().reset_index()

subset 
```

```python
plt.plot(
    subset['education'],
    subset['fnlwgt']
)
plt.title('Amount of Population w / Final Eduucation Level')
plt.xlabel('Education Level')
plt.ylabel('Population (Billions)')
plt.xticks(rotation=90)
plt.grid(alpha=0.3)
plt.show
```
